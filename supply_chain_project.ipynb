```python
data_clean = data_base.copy()
```

```python
numeric_comma_cols = [
    'Avg_Daily_Sales',
    'Days_of_Inventory',
    'SKU_Churn_Rate',
    'Order_Frequency_per_month'
]

for col in numeric_comma_cols:
    data_clean[col] = (
        data_clean[col].astype(str)  # Explicitly convert to string type
        .str.replace('.', '', regex=False)  # Remove thousands separator
        .str.replace(',', '.', regex=False)  # Replace decimal comma with period
        .astype(float)
    )
```

```python
currency_cols = [
    'Unit_Cost_USD',
    'Last_Purchase_Price_USD',
    'Total_Inventory_Value_USD'
]

for col in currency_cols:
    data_clean[col] = (
        data_clean[col]
        .str.replace('$', '', regex=False)
        .str.replace('.', '', regex=False)
        .str.replace(',', '.', regex=False)
        .astype(float)
    )
```

```python
percentage_cols = [
    'Supplier_OnTime_Pct',
    'Audit_Variance_Pct',
    'Demand_Forecast_Accuracy_Pct'
]

for col in percentage_cols:
    data_clean[col] = (
        data_clean[col]
        .str.replace('%', '', regex=False)
        .str.replace(',', '.', regex=False)
        .astype(float)
    )
```

## **FEATURE EXTRACTION**

```python
#ATP=Quantity_On_Hand−Quantity_Reserved−Quantity_Committed
data_clean['Available_to_Promise'] = (
    data_clean['Quantity_On_Hand']
    - data_clean['Quantity_Reserved']
    - data_clean['Quantity_Committed']
)
```

```python
data_clean['Forecast_Next_30d'] = data_base['Forecast_Next_30d']
```

```python
data_clean['Forecast_Daily_Demand'] = (
    data_clean['Forecast_Next_30d'] / 30
)
```

```python
data_clean['Forecast_Error'] = (
    data_clean['Forecast_Daily_Demand']
    - data_clean['Avg_Daily_Sales']
)
```

```python
data_clean['Forecast_Error_Pct'] = (
    data_clean['Forecast_Error']
    / data_clean['Avg_Daily_Sales']
) * 100
```

```python
data_clean[
    [
        'SKU_ID',
        'Avg_Daily_Sales',
        'Forecast_Next_30d',
        'Forecast_Daily_Demand',
        'Forecast_Error',
        'Forecast_Error_Pct'
    ]
].head(10)
```

```python
data_clean['Calculated_Inventory_Value'] = (
    data_clean['Quantity_On_Hand'] *
    data_clean['Unit_Cost_USD']
)
```

```python
data_clean['Inventory_Value_Difference'] = (
    data_clean['Total_Inventory_Value_USD']
    - data_clean['Calculated_Inventory_Value']
)
```

```python
data_clean[
    [
        'SKU_ID',
        'Quantity_On_Hand',
        'Unit_Cost_USD',
        'Total_Inventory_Value_USD',
        'Calculated_Inventory_Value',
        'Inventory_Value_Difference'
    ]
].head(10)
```

```python
data_clean['Inventory_Value_Diff_Pct'] = (
    abs(data_clean['Inventory_Value_Difference'])
    / data_clean['Calculated_Inventory_Value']
) * 100

data_clean[
    data_clean['Inventory_Value_Diff_Pct'] > 5
][[
    'SKU_ID',
    'Quantity_On_Hand',
    'Unit_Cost_USD',
    'Total_Inventory_Value_USD',
    'Calculated_Inventory_Value',
    'Inventory_Value_Difference',
    'Inventory_Value_Diff_Pct'
]]
```

# REORDER RISK

```python
data_clean['Reorder_Gap'] = (
    data_clean['Quantity_On_Hand']
    - data_clean['Reorder_Point']
)
```

```python
data_clean['Reorder_Risk'] = (
    data_clean['Quantity_On_Hand']
    < data_clean['Reorder_Point']
)
```

```python
print(
    data_clean['Reorder_Risk'].value_counts()
)
```

```python
print(
    "Reorder Risk %:",
    round(data_clean['Reorder_Risk'].mean() * 100, 2)
)
```

```python
data_clean[
    [
        'SKU_ID',
        'Quantity_On_Hand',
        'Reorder_Point',
        'Safety_Stock',
        'Reorder_Gap',
        'Reorder_Risk'
    ]
].head(10)
```

```python
data_clean['Safety_Stock_Gap'] = (
    data_clean['Quantity_On_Hand']
    - data_clean['Safety_Stock']
)
```

```python
data_clean['Safety_Stock_Risk'] = (
    data_clean['Quantity_On_Hand']
    < data_clean['Safety_Stock']
)
```

```python
print(data_clean['Safety_Stock_Risk'].value_counts())

print(
    "Safety Stock Risk %:",
    round(data_clean['Safety_Stock_Risk'].mean() * 100, 2)
)
```

```python
risk_summary = pd.DataFrame({
    'Risk_Metric': [
        'Reorder Risk',
        'Safety Stock Risk',
        'Below Lead Time Coverage'
    ],
    'SKU_Count': [
        data_clean['Reorder_Risk'].sum(),
        data_clean['Safety_Stock_Risk'].sum(),
        (data_clean['Stock_Coverage_vs_LeadTime'] < 0).sum()
    ]
})

risk_summary['Risk_Pct'] = (
    risk_summary['SKU_Count']
    / len(data_clean)
    * 100
).round(2)

risk_summary
```

```python
data_clean['Inventory_Risk_Score'] = (
    data_clean['Reorder_Risk'].astype(int)
    + data_clean['Safety_Stock_Risk'].astype(int)
    + (data_clean['Stock_Coverage_vs_LeadTime'] < 0).astype(int)
)
```

```python
risk_distribution = (
    data_clean['Inventory_Risk_Score']
    .value_counts()
    .sort_index()
    .reset_index()
)

risk_distribution.columns = ['Risk_Score', 'SKU_Count']

risk_distribution['Risk_Pct'] = (
    risk_distribution['SKU_Count']
    / len(data_clean)
    * 100
).round(2)

risk_distribution
```

```python
pd.crosstab(
    data_clean['Inventory_Risk_Score'],
    data_clean['Inventory_Status']
)
```

```python
category_analysis = (
    data_clean
    .groupby('Category')
    .agg(
        SKUs=('SKU_ID', 'nunique'),
        Inventory_Units=('Quantity_On_Hand', 'sum'),
        Inventory_Value=('Total_Inventory_Value_USD', 'sum'),
        Avg_Days_Inventory=('Days_of_Inventory', 'mean'),
        Low_Stock_SKUs=('Reorder_Risk', 'sum'),
        Negative_ATP_SKUs=('Available_to_Promise', lambda x: (x < 0).sum())
    )
    .sort_values('Inventory_Value', ascending=False)
)

category_analysis
```

```python
category_analysis['Inventory_Value_Pct'] = (
    category_analysis['Inventory_Value']
    / data_clean['Total_Inventory_Value_USD'].sum()
    * 100
).round(2)

category_analysis
```

```python
warehouse_analysis = (
    data_clean
    .groupby(['Warehouse_ID', 'Warehouse_Location'])
    .agg(
        SKUs=('SKU_ID', 'nunique'),
        Inventory_Units=('Quantity_On_Hand', 'sum'),
        Inventory_Value=('Total_Inventory_Value_USD', 'sum'),
        Avg_Days_Inventory=('Days_of_Inventory', 'mean'),
        Reorder_Risk_SKUs=('Reorder_Risk', 'sum'),
        Negative_ATP_SKUs=(
            'Available_to_Promise',
            lambda x: (x < 0).sum()
        )
    )
    .sort_values('Inventory_Value', ascending=False)
)

warehouse_analysis
```

```python
warehouse_analysis['Inventory_Value_Pct'] = (
    warehouse_analysis['Inventory_Value']
    / data_clean['Total_Inventory_Value_USD'].sum()
    * 100
).round(2)

warehouse_analysis
```

```python
supplier_analysis = (
    data_clean
    .groupby('Supplier_Name')
    .agg(
        SKUs=('SKU_ID', 'nunique'),
        Inventory_Value=('Total_Inventory_Value_USD', 'sum'),
        Avg_OnTime=('Supplier_OnTime_Pct', 'mean'),
        Avg_Lead_Time=('Lead_Time_Days', 'mean'),
        Reorder_Risk=('Reorder_Risk', 'sum')
    )
    .sort_values('Avg_OnTime')
)

supplier_analysis
```
